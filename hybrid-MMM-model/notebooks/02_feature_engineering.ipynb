```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/cleaned_marketing_data.csv"
)

df["DATE_DAY"] = pd.to_datetime(df["DATE_DAY"])
```

```python
# Create Log Revenue
df["LOG_REVENUE"] = np.log1p(
    df["ALL_PURCHASES_ORIGINAL_PRICE"]
)
```

```python
# Create Total Clicks 

click_cols = [
    "GOOGLE_PAID_SEARCH_CLICKS",
    "GOOGLE_SHOPPING_CLICKS",
    "GOOGLE_PMAX_CLICKS",
    "GOOGLE_DISPLAY_CLICKS",
    "GOOGLE_VIDEO_CLICKS",
    "META_FACEBOOK_CLICKS",
    "META_INSTAGRAM_CLICKS",
    "META_OTHER_CLICKS",
    "TIKTOK_CLICKS"
]

df[click_cols] = df[click_cols].fillna(0)

df["TOTAL_CLICKS"] = df[
    click_cols
].sum(axis=1)
```

```python
# Create Impression

imp_cols = [
    "GOOGLE_PAID_SEARCH_IMPRESSIONS",
    "GOOGLE_SHOPPING_IMPRESSIONS",
    "GOOGLE_PMAX_IMPRESSIONS",
    "GOOGLE_DISPLAY_IMPRESSIONS",
    "GOOGLE_VIDEO_IMPRESSIONS",
    "META_FACEBOOK_IMPRESSIONS",
    "META_INSTAGRAM_IMPRESSIONS",
    "META_OTHER_IMPRESSIONS",
    "TIKTOK_IMPRESSIONS"
]

df[imp_cols] = df[imp_cols].fillna(0)

df["TOTAL_IMPRESSIONS"] = df[
    imp_cols
].sum(axis=1)
```

```python
# Create ctr
df["CTR"] = (
    df["TOTAL_CLICKS"]
    /
    (df["TOTAL_IMPRESSIONS"] + 1)
)
```

```python
# Create Total Purchase

df["TOTAL_PURCHASES"] = (
    df["ALL_PURCHASES"]
)
```

```python
# Create ROAS (Return on Ad Spend)
df["ROAS"] = np.where(
    df["TOTAL_MARKETING_SPEND"] > 0,
    df["ALL_PURCHASES_ORIGINAL_PRICE"] /
    df["TOTAL_MARKETING_SPEND"],
    np.nan
)
```

```python
df["ROAS"].describe()
```

```python
# Create Digit Spend
df["DIGITAL_SPEND"] = (
      df["GOOGLE_PAID_SEARCH_SPEND"]
    + df["GOOGLE_SHOPPING_SPEND"]
    + df["GOOGLE_PMAX_SPEND"]
    + df["GOOGLE_DISPLAY_SPEND"]
    + df["GOOGLE_VIDEO_SPEND"]
    + df["META_FACEBOOK_SPEND"]
    + df["META_INSTAGRAM_SPEND"]
    + df["META_OTHER_SPEND"]
    + df["TIKTOK_SPEND"]
)
```

```python
# Checking Features
features = [
    "TOTAL_MARKETING_SPEND",
    "TOTAL_CLICKS",
    "TOTAL_IMPRESSIONS",
    "CTR",
    "ROAS",
    "DIGITAL_SPEND",
    "LOG_REVENUE"
]

df[features].head()
```
