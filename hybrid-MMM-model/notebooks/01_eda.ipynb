```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/marketing_data.csv")
```

```python
print(df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
#Missing Value Analysis

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": round(df.isnull().mean()*100,2)
})

missing.sort_values(
    by="Missing_Percentage",
    ascending=False
)
```

```python
# Convert data
df["DATE_DAY"] = pd.to_datetime(df["DATE_DAY"])
```

```python
df["DATE_DAY"].head()
```

```python
# Time Features
df["YEAR"] = df["DATE_DAY"].dt.year

df["MONTH"] = df["DATE_DAY"].dt.month

df["QUARTER"] = df["DATE_DAY"].dt.quarter

df["DAY_OF_WEEK"] = df["DATE_DAY"].dt.dayofweek
```

```python
df[
    [
        "DATE_DAY",
        "YEAR",
        "MONTH",
        "QUARTER",
        "DAY_OF_WEEK"
    ]
].head()
```

```python
# Marketing Spend Columns

spend_cols = [
    "GOOGLE_PAID_SEARCH_SPEND",
    "GOOGLE_SHOPPING_SPEND",
    "GOOGLE_PMAX_SPEND",
    "GOOGLE_DISPLAY_SPEND",
    "GOOGLE_VIDEO_SPEND",
    "META_FACEBOOK_SPEND",
    "META_INSTAGRAM_SPEND",
    "META_OTHER_SPEND",
    "TIKTOK_SPEND"
]
```

```python
df[spend_cols] = df[spend_cols].fillna(0)
```

```python
df[spend_cols].isnull().sum()
```

```python
# Creating Total Market Spend
df["TOTAL_MARKETING_SPEND"] = df[
    spend_cols
].sum(axis=1)
```

```python
df[
    [
        "TOTAL_MARKETING_SPEND"
    ]
].head()
```

```python
# Revenue Variable

target = "ALL_PURCHASES_ORIGINAL_PRICE"
```

```python
df[target].describe()
```

```python
# Revenue Distribution 

plt.figure(figsize=(10,5))

plt.hist(
    df[target],
    bins=50
)

plt.title(
    "Revenue Distribution"
)

plt.show()
```

```python
# Total Spend vs Revenue 

plt.figure(figsize=(10,6))
plt.scatter(
    df["TOTAL_MARKETING_SPEND"],
    df[target],
    alpha=0.3)
plt.xlabel("Total Marketing Spend")
plt.ylabel("Revenue")
plt.title("Marketing Spend vs Revenue")
plt.show()
```

```python
# Correlation Analysis

corr_df = df[
    spend_cols + [target]
]
```

```python
corr = corr_df.corr()
```

```python
corr[target].sort_values(ascending=False)
```

```python
# Heat Map
plt.figure(figsize=(12,8))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```
